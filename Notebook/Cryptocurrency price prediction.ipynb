## 1. Bitcoin LSTM Model

```python
from sklearn.preprocessing import MinMaxScaler
from sklearn.model_selection import train_test_split

from keras.models import Sequential
from keras.layers import LSTM, Dense, Dropout
```

```python
btc_data = btc[["Close"]]
btc_scaler = MinMaxScaler(feature_range=(0,1))
btc_scaled = btc_scaler.fit_transform(btc_data)
```

```python
print(btc_scaled[:5])
```

```python
import joblib
joblib.dump(btc_scaler, "btc_scaler.pkl")
```

```python
X_btc = []
y_btc = []

time_step = 60

for i in range(time_step, len(btc_scaled)):
    X_btc.append(btc_scaled[i-time_step:i, 0])
    y_btc.append(btc_scaled[i, 0])

X_btc = np.array(X_btc)
y_btc = np.array(y_btc)
```

```python
X_btc = X_btc.reshape(X_btc.shape[0],X_btc.shape[1],1)

print(X_btc.shape)
print(y_btc.shape)
```

```python
X_train_btc, X_test_btc, y_train_btc, y_test_btc = train_test_split(
    X_btc,
    y_btc,
    test_size=0.3,
    shuffle=False
)
```

```python
model_btc = Sequential()

model_btc.add(LSTM(units=50, return_sequences=True, input_shape=(X_train_btc.shape[1], 1)))
model_btc.add(Dropout(0.2))

model_btc.add(LSTM(units=50))
model_btc.add(Dropout(0.2))

model_btc.add(Dense(units=25))
model_btc.add(Dense(units=1))
```

```python
model_btc.compile(
    optimizer='adam',
    loss='mean_squared_error'
)
```

```python
model_btc.summary()
```

```python
model_btc.fit(
    X_train_btc,
    y_train_btc,
    epochs=30,
    batch_size=32,
    validation_data=(X_test_btc,y_test_btc)
)
```

```python
model_btc.save("bitcoin_model.keras")
```

```python
y_pred_scaled = model_btc.predict(X_test_btc)
```

```python
y_pred = btc_scaler.inverse_transform(y_pred_scaled)

y_test_actual = btc_scaler.inverse_transform(
    y_test_btc.reshape(-1, 1)
)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

mae = mean_absolute_error(y_test_actual, y_pred)
rmse = np.sqrt(mean_squared_error(y_test_actual, y_pred))
r2 = r2_score(y_test_actual, y_pred)

print("MAE:", mae)
print("RMSE:", rmse)
print("R² Score:", r2)
```

```python
last_60 = btc[['Close']].tail(60)

X = btc_scaler.transform(last_60)
X = X.reshape(1, 60, 1)

prediction = model_btc.predict(X)
next_price = btc_scaler.inverse_transform(prediction)

print("Next Day Bitcoin Price (USD):", next_price[0][0])
```

```python
print("Last Closing Price (USD):", btc['Close'].iloc[-1])
print("Predicted Next Day Price (USD):", next_price[0][0])
```

```python
test_dates = btc['Date'].iloc[-len(y_test_actual):]
```

```python
plt.figure(figsize=(12,6))
plt.plot(test_dates,y_test_actual, label="Actual Price")
plt.plot(test_dates,y_pred, label="Predicted Price")
plt.title("Bitcoin Price Prediction")
plt.xlabel("Date")
plt.ylabel("Price (USD)")
plt.legend()
plt.show()
```
